```python
# Imports python libraries

import numpy as np
import matplotlib.pyplot as plt
```

```python
#Uploads the four data files as numpy arrays.
spikes_left_trial1 = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_left_trial1.csv', delimiter=',')
spikes_right_trial1 = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_right_trial1.csv', delimiter=',')
spikes_left_trial2 = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_left_trial2.csv', delimiter=',')
spikes_right_trial2 = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_right_trial2.csv', delimiter=',')
```

```python
#Prints the length of each array.
print(f'Length of the Left Trial 1 array is: {len(spikes_left_trial1)}.')
print(f'Length of the Right Trial 1 array is: {len(spikes_right_trial1)}.')
print(f'Length of the Left Trial 2 array is: {len(spikes_left_trial2)}.')
print(f'Length of the Right Trial 2 array is: {len(spikes_right_trial2)}.')
```

```python
#Determines the firing rate of the neuron during each trial.
rate_left_trial1 = len(spikes_left_trial1) / spikes_left_trial1[-1]
print(f'The firing rate of the Left Trial 1 array is: {rate_left_trial1}.')

rate_right_trial1 = len(spikes_right_trial1) / spikes_right_trial1[-1]
print(f'The firing rate of the Right Trial 1 array is: {rate_right_trial1}.')

rate_left_trial2 = len(spikes_left_trial2) / spikes_left_trial2[-1]
print(f'The firing rate of the Left Trial 2 array is: {rate_left_trial2}.')

rate_right_trial2 = len(spikes_right_trial2) / spikes_right_trial2[-1]
print(f'The firing rate of the Right Trial 2 array is: {rate_right_trial2}.')
```

It is a little hard to say if the left or right trials have the highest firing rates because the first right trial holds to lowest rate while the second right trial holds the highest rate.

However, if we average the left trial rates (1.9845) and the right trial rates (2.0712), respectively, we can find that the neurons has a slightly higher firing rate before the right turn trials.

```python
#Creates the subplots
fig, ax = plt.subplots(2, 1, figsize=(10, 6))

#Creates the graph for the first trials of both the left and right turns.
ax[0].scatter(spikes_left_trial1, np.ones_like(spikes_left_trial1)) #Plots points in a line
ax[0].scatter(spikes_right_trial1, np.ones_like(spikes_right_trial1)*2) #Plots points in a line above the first
ax[0].set_title('Trial 1 Spike Train')
ax[0].set_xlabel('Time (s)')
ax[0].legend({'Left', 'Right'})
ax[0].set_yticks([]) #Removes the y-ticks
ax[0].set_ylim([0,3]) #Sets y-axis limits

#Creates the graph for the second trials of both the left and right turns.
ax[1].scatter(spikes_left_trial2, np.ones_like(spikes_left_trial2))
ax[1].scatter(spikes_right_trial2, np.ones_like(spikes_right_trial2)*2)
ax[1].set_title('Trial 2 Spike Train')
ax[1].set_xlabel('Time (s)')
ax[1].legend({'Left', 'Right'})
ax[1].set_yticks([])
ax[1].set_ylim([0,3])

#Plot the graphs!
plt.tight_layout() #Found this on google just to fix the layout of the graphs so that the x-axis label from the top graph and the title from the bottom graph do not overlap...
plt.show()
```

The neuron seems to have a more "bursty" firing on both trials and for both the left and the right, since we see frequent bursts of firings with breaks in between.

That said, it seems that the left turn trials did induce more "bursty" firing patterns from the neuron while the right turn trials showed slightly more regular firing patterns. We see this because there are more "clumps" of firings with bigger gap breaks in the left turn trials while the firings were a bit more spread out in the right turn trials.



```python
#Calculates the ISI arrays for each of the four skip arrays. (and print)
ISI_left_trial1 = np.diff(spikes_left_trial1)
print(f'The ISI of the Left Trial 1 array is:')
print(f'{ISI_left_trial1}')

ISI_right_trial1 = np.diff(spikes_right_trial1)
print(f'The ISI of the right Trial 1 array is:')
print(f'{ISI_right_trial1}')

ISI_left_trial2 = np.diff(spikes_right_trial2)
print(f'The ISI of the Left Trial 2 array is:')
print(f'{ISI_left_trial2}')

ISI_right_trial2 = np.diff(spikes_right_trial2)
print(f'The ISI of the right Trial 2 array is:')
print(f'{ISI_right_trial2}')
```

```python
#Defines the bins for the histogram.
bins = np.arange(0, 2, 0.01)

#Creates the subplots.
fig, ax = plt.subplots(2, 2, figsize=(12, 8))

#Creates the histogram with labeled axes for the first trial of left turns.
ax[0, 0].hist(ISI_left_trial1, bins=bins)
ax[0, 0].set_title('ISI Left Trial 1')
ax[0, 0].set_ylabel('Frequency')
ax[0, 0].set_xlabel('Time (s)')

#Creates the histogram with labeled axes for the first trial of right turns.
ax[0, 1].hist(ISI_right_trial1, bins=bins)
ax[0, 1].set_title('ISI Right Trial 1')
ax[0, 1].set_ylabel('Frequency')
ax[0, 1].set_xlabel('Time (s)')

#Creates the histogram with labeled axes for the second trial of left turns.
ax[1, 0].hist(ISI_left_trial2, bins=bins)
ax[1, 0].set_title('ISI Left Trial 2')
ax[1, 0].set_ylabel('Frequency')
ax[1, 0].set_xlabel('Time (s)')

#Creates the histogram with labeled axes for the second trial of right turns.
ax[1, 1].hist(ISI_right_trial2, bins=bins)
ax[1, 1].set_title('ISI Right Trial 2')
ax[1, 1].set_ylabel('Frequency')
ax[1, 1].set_xlabel('Time (s)')

#Plot the histograms!
plt.tight_layout() #Like before... found this on google just to fix the layout of the graphs so that the x-axis label from the top graph and the title from the bottom graph do not overlap.
plt.show()
```

From these histograms, it does look like this neuron fires in bursts. Particularly, the neuron has a high "burstiness" during the begining of each trial.

That said, similarly to the dot plot graphs, I do believe that the neurons has slightly higher "burstiness" during the left turn trials since there are more visible clumps and gaps on the histograms.

```python
#Defines time bins.
time_bins = np.arange(0, 31, 1) #Put 31 for end time so that there will be an x-axis tick of "30"

#Computes histograms of the data sets in accordance with the time bins.
FR_over_time_left_trial1, _ = np.histogram(spikes_left_trial1, time_bins)
FR_over_time_right_trial1, _ = np.histogram(spikes_right_trial1, time_bins)
FR_over_time_left_trial2, _ = np.histogram(spikes_left_trial2, time_bins)
FR_over_time_right_trial1, _ = np.histogram(spikes_right_trial2, time_bins)

#Creates the subplots.
fig, ax = plt.subplots(2, 1, figsize=(10, 8))

#Creates the graph with labeled axes for the firing frequency over time for the neuron during the first trial of left turns.
ax[0].plot(FR_over_time_left_trial1)
ax[0].plot(FR_over_time_right_trial1)
ax[0].set_title('Trial 1 Firing Rate Over Time')
ax[0].set_ylabel('Firing Rate (Hz)')
ax[0].set_xlabel('Time (s)')
ax[0].legend({'Left', 'Right'})

#Creates the graph with labeled axes for the firing frequency over time for the neuron during the second trial of left turns.
ax[1].plot(FR_over_time_left_trial2)
ax[1].plot(FR_over_time_right_trial1)
ax[1].set_title('Trial 2 Firing Rate Over Time')
ax[1].set_ylabel('Firing Rate (Hz)')
ax[1].set_xlabel('Time (s)')
ax[1].legend({'Left', 'Right'})

#Plot the histograms!
plt.tight_layout()
plt.show()
```

Looking at the graphs, it is quite apparents that the pattens in firing rates over time are very different both between the left and right turn trials as well as the first and second left turn trials. However, the pattern of firing rates over time between the first and second right turn trials.

In the first trials, we see that the left turn trial induced lower firing rates throughout the trial, with a high firing rate apike around the 15s mark. The right turn trial, similarly to the right turn trial 2, has the highest firing rate at the beginning of the trials in addition to a couple spikes around the 9 and 15s marks. The Left turn second trial, unline the left turn first trial and the right turn trials, has a couple small spikes around the 4 and 11s marks but high higher firing rate spikes between the 20-20s time marks.
