```python
# Imports relevant python libraries

import numpy as np
import matplotlib.pyplot as plt
```

```python
# Imports the acc_hits_hw5.csv and acc_fa_hw5.csv files as numpy arrays

acc_hits = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/acc_hits_hw5.csv', delimiter = ',')
acc_fa = np.loadtxt('https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/acc_fa_hw5.csv', delimiter = ',')
```

```python
#Prints the shape of our acc_hits and acc_fa arrays

print(f'The shape of the acc_hits array is {acc_hits.shape}')
print(f'The shape of the acc_fa array is {acc_fa.shape}')
```

```python
#Creats a time variable that has the same length as the acc_hits and acc_fa arrays, with equally spaced values between 'zero' and 'four'

time = np.linspace(0, 4, 8001)
time
```

```python
#Finds the sampling rate of the data

sampling_rate = (8001/4)
sampling_rate
```

```python
#Plots the first trials of acc_hits and acc_fa over time in two respective subplots.

#Sets up the subplots
fig, ax = plt.subplots(2,1)

#Plots the first trial of acc_hits over time. Also sets axis labels and title
ax[0].plot(time, acc_hits[0])
ax[0].set_title('acc_hits Trial 1 Over Time')
ax[0].set_xlabel('Time [s]')
ax[0].set_ylabel('Voltage')

#Plots the first trial of acc_fa over time. Also sets axis labels, and title
ax[1].plot(time, acc_fa[0])
ax[1].set_title('acc_fa Trial 1 Over Time')
ax[1].set_xlabel('Time [s]')
ax[1].set_ylabel('Voltage')

#Actually plots the graph
plt.tight_layout() #Found online and used in previous notebooks to make the graphs look better
plt.show()
```

From the top graph, it looks like the dominant oscillation for this LFP is 1.25 Hz. This is because I can see 5 peaks in the graph of the acc_hits first trial over 4 seconds of measurement. So, 5/4 = 1.25!

Looking at the bottom graph, I do not believe that the dominant oscillation is different ebtween the first trial of acc_hit and the first trial of acc_fa. In the acc_fa graph, I am also able to recognize five(one cut off) peaks in the 4 seconds of measurement. So, for that graph too, I would say that the dominant oscillation is 1.25.

```python
#Calculates and plots the ERP for acc_hits

#Calculates and plots the ERP
plt.plot(time, acc_hits.mean(0))

#Labels axes and sets title
plt.xlabel('Time [s]')
plt.ylabel('Voltage')
plt.title('ERP of acc_hits')

#Adds vertical and horizontal lines to the graph
plt.vlines(2, 54, -50, color = 'black')
plt.hlines(0, 0, 4, color = 'black')
```

No, it does not really look like there is consistent activity in the LFP surrounding the screen touch. However, the activity does look very different before and after the screen touch. Before the screen touch, or on the left of the vertical line, there is very small activity with a few weak peaks and troughs. After the screen touch, on the right of the vertical line, there seems to be much stronger activity but with fewer peaks and troughs.

```python
#Calculates and plots the ERP for acc_fa (***REPEATING STEP 10 FOR acc_fa***)

#Calculates and plots the ERP
plt.plot(time, acc_fa.mean(0))

#Labels axes and sets title
plt.xlabel('Time [s]')
plt.ylabel('Voltage')
plt.title('ERP of acc_fa')

#Adds vertical and horizontal lines to the graph
plt.vlines(2, 54, -50, color = 'black')
plt.hlines(0, 0, 4, color = 'black')
```

```python
#Calculates and plots the ERP for acc_hits and acc_fa on the same graph

#Calculates and plots the ERP of both acc_hits and acc_fa
plt.plot(time, acc_hits.mean(0), color = 'red')
plt.plot(time, acc_fa.mean(0), color = 'blue')

#Labels axes and sets title
plt.xlabel('Time [s]')
plt.ylabel('Voltage')
plt.title('ERP of acc_hits and acc_fa')

#Adds vertical and horizontal lines to the graph
plt.vlines(2, 54, -50, color = 'black')
plt.hlines(0, 0, 4, color = 'black')

plt.show()
```

The graphs of the ERPs of acc_hits(red) and acc_fa(blue) are quite different.

The graph showing the ERP of acc_fa shows consitent activity both before and after the touch screen (around the vertical line), but this activity is always rather weak. The two graphs are similar to the left of the vertical line, though their sign (+ or -) does seem to be opposites. On the other hand, to the right of the vertical line, the acc_hits graphs shows much stringer activity which the acc_fa graph remains the same.
